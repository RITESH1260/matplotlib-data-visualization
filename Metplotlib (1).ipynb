```python
#Here we are discuss about matplotlib, pandas and random module

import matplotlib.pyplot as plt
import pandas as pd
import random 
```

```python
#Making a chart using pandas data

data = {
    "Month" : ["jan","Feb","Mar", "Apr", "May"],
    "Sales" : [12000, 25000, 16000, 31000, 18000]
}

df = pd.DataFrame(data)
df
```

```python
plt.bar(df["Month"], df["Sales"])

plt.title("Pandas Data")
plt.style.use("dark_background")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
#Here we use subplot for the making a two or multiple charts at a time

category1 = ["car", "scooter", "bus", "truck", "aeroplane"]
sale1 = [ 20, 56, 45, 68, 25]

category2 = ["car", "scooter", "bus", "truck", "aeroplane"]
sale2 = [45, 26 , 50, 23, 42]

plt.figure(figsize= (15,5))
plt.subplot(1,2,1)
plt.bar(category1, sale1)
plt.title("subplot data")
plt.xlabel("category")
plt.ylabel("sale")


plt.subplot(1,2,2)
plt.bar(category2, sale2)
plt.title("subplot data 2")
plt.xlabel("category")
plt.ylabel("sale")
plt.savefig("pandas_data.png")

plt.show()

 
```

```python
import matplotlib.pyplot as plt
import numpy as np

species = ("Adenine", "Chinstrap", "Gentoo")
penguin_means = {
    "Bill Depth" : (18.35, 18.43, 14.9),
    "Bill Length" : (38.39, 48.83, 47.35),
    "Flipper Length" : (189.35, 195.32, 207.32)
}

x= np.arange(len(species)) #the label location
width = 0.25 #width of bars
multiplier = 0

fig, ax = plt.subplots(layout = "constrained")

for attribute, measurement in penguin_means.items():
    offset = width * multiplier
    react = ax.bar(x + offset, measurement, width, label = attribute)
    ax.bar_label(react, padding = 3) #For seen a data value on each bar 
    multiplier += 1

#Add some textfor lebels, title andcustom x-axis tick label, etc.
ax.set_ylabel("length(mm)")
ax.set_title("penguins attribute byspecies")
ax.set_xticks(x + width, species)
ax.legend(loc = "upper left", ncols = 3)
ax.set_ylim(0,250)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

t = np.linspace(-10, 10, 100)
sig = 1/(1 + np.exp(-t))

fig, ax = plt.subplots()
ax.axhline(y = 0, color = "black", linestyle = "--")
ax.axhline(y = 0.5, color = "black", linestyle = "--")
ax.axhline(y = 1.0, color = "black", linestyle = "--")
ax.axvline(color = "grey")
ax.axline((0, 0.5), slope = .25, color = "black", linestyle = "--")
ax.plot(t, sig, linewidth = 2, label = r"$\sigma(t)= \frac{1}{1 + e^{-t}}$")
ax.set(xlim= (-10, 10), xlabel = "t")
ax.legend(fontsize= 14)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Fixing random state for reproducibility
np.random.seed(19680801)

dt = 0.01
t = np.arange(0, 30, dt)
nse1 = np.random.randn(len(t))
nse2 = np.random.randn(len(t))

#Two signals with a coherent part at 10 Hz and a random part
s1 = np.sin(2 * np.pi *10 * t) + nse1
s2 = np.sin(2 * np.pi * 10* t) + nse2

fig, axs = plt.subplots(2, 1 , layout= "constrained")
axs[0].plot(t, s1, t, s2)
axs[0].set_xlim(0, 2)
axs[0].set_xlabel("Time(s)")
axs[0].set_ylabel("s1 and s2")
axs[0].grid(True)

cxy, f = axs[1].cohere(s1, s2, NFFT = 256, Fs = 1. / dt)
axs[1].set_ylabel("Coherence")

plt.show()
```
