```python
#Here we are discuss about matplotlib, pandas and random module

import matplotlib.pyplot as plt
import pandas as pd
import random 
```

```python
#Making a chart using pandas data

data = {
    "Month" : ["jan","Feb","Mar", "Apr", "May"],
    "Sales" : [12000, 25000, 16000, 31000, 18000]
}

df = pd.DataFrame(data)
df
```

```python
plt.bar(df["Month"], df["Sales"])

plt.title("Pandas Data")
plt.style.use("dark_background")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
#Here we use subplot for the making a two or multiple charts at a time

category1 = ["car", "scooter", "bus", "truck", "aeroplane"]
sale1 = [ 20, 56, 45, 68, 25]

category2 = ["car", "scooter", "bus", "truck", "aeroplane"]
sale2 = [45, 26 , 50, 23, 42]

plt.figure(figsize= (15,5))
plt.subplot(1,2,1)
plt.bar(category1, sale1)
plt.title("subplot data")
plt.xlabel("category")
plt.ylabel("sale")


plt.subplot(1,2,2)
plt.bar(category2, sale2)
plt.title("subplot data 2")
plt.xlabel("category")
plt.ylabel("sale")
plt.savefig("pandas_data.png")

plt.show()

 
```

```python
#For a random Value, we use histogram chart 


dict =[random.randint(1,10) for i in range(0, 100)]
plt.hist(dict, bins = 10)
plt.title("Random value using")
plt.show()
```

```python
#Scatter plot
x = [10, 20, 30,40,50]
y = [20, 30, 70, 55, 60]
plt.scatter(x,y)
plt.title("Scatter plot")
plt.show()
```

```python
#Point graph
x = [25, 42, 56, 23, 45]
y = [52, 42, 51, 32, 54]

plt.plot(x,y)
plt.title("Point plot")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.show()
```

```python
x = ['A', 'B', 'C', 'D','E','F','G']
y = [10, 20, 15, 55, 45, 10, 25]

plt.pie(y, labels = x, autopct = "%1.1f%%")
plt.title("Pie Chart")
plt.show()
```
